## Naming Rules

Variables can be any name except the reserved keywords (see below) and
variables must start with a letter or _ . <br>
Variables cannot start with a number or special character.
Do not use special characters like +,-,*,/,==, !, @, % or similar. These are operators or have specific meanings in the code.
It is best to stay with combinations of capital/lowercase letters(A, a), numbers(1,2,3) and underscores "_".
Keep the variable names short, it saves time and reduces the chances of errors.

The following are examples of acceptable names:
- X
- x
- abc
- _1ABC
- Input_Data


The following are not acceptable names: <br>
- 01_red - begins with a number
- #unit - begins with a special character, #,  which means comment not code
- class - begins with a keyword
- True, False - begins with keywords words
- input data - variable names cannot have spaces in them

To see reserved keywords type help("keywords")<br>

*The help function is useful when you need to check syntax, check parameters or sub functions.  We will use this occasionally in the course.*

```python
help('keywords')
```

## RABBIT HOLE  # 1

For more informaion on keywords and what they do see.
  
For the latest updates copy and run the following code block.<br>
The video below gives a nice explanation of the function of the various keywords. <br>
Link to Indently video on YouTube https://youtu.be/rKk8XPLysj8?si=TitMkWiBnsZQUIrn 



```python
from keyword import kwlist, softkwlist

def display_keywords():
    print('Keywords:')
    for i, kw in enumerate(kwlist, start=1):
        print(f'{i:2}: {kw}')
    print('Soft keywords:')
    for i, skw in enumerate(softkwlist, start=1):
        print(f'{i:2}: {skw}')

display_keywords()
```

### Mathematical operations<br>

## Logical statements and operators

```python
#In basic Python, we check likeness.
x = 5  # assign x the value of 5
if x>y:
    print(x)
else:
    print(x, " Is not greater than " , y)
```

 Logical statements take the form of : <br>
        If x is True/False do:<br>
           - something<br>
        else:<br>
           - do something else.<br>

You will also see the term Boolean (Boole, an English mathematician in the 19th century). This is a special type of algebra concerned with binary variables and logical operations. It has operators **and**, **not**, and **or**, and returns **True** or **False** results.
